##  Contexto Teórico

### Por que isso importa?

O trabalho de jovens durante os estudos é uma realidade comum no Brasil, especialmente entre estudantes de baixa renda. Compreender como essa prática afeta o desempenho educacional é crucial para:

1. **Políticas Educacionais**: Identificar se programas de bolsas/auxílio podem reduzir a necessidade de trabalho.
2. **Combate à Desigualdade**: Verificar se o trabalho amplifica desigualdades socioeconômicas.
3. **Orientação Vocacional**: Ajudar estudantes a equilibrar trabalho e estudos.
4. **Legislação Trabalhista**: Embasar discussões sobre jornadas de trabalho para jovens.

### Hipóteses Iniciais

**H1:** Estudantes que trabalham têm desempenho médio inferior aos que não trabalham.  
**H2:** Quanto maior a carga horária de trabalho, menor o desempenho.  
**H3:** O impacto negativo do trabalho é maior para estudantes de baixa renda.  
**H4:** Trabalhar período integral reduz significativamente as notas em todas as áreas.  
**H5:** O efeito do trabalho é mais pronunciado em regiões com menor infraestrutura educacional.

##  Contextualização Nacional

### Dados do Mercado de Trabalho Jovem no Brasil

- **IBGE (2023)**: Aproximadamente **40% dos jovens de 18-24 anos** conciliam trabalho e estudos.
- **PNAD Contínua**: Jovens de famílias com renda até 2 salários mínimos têm **3x mais chances** de trabalhar durante os estudos.
- **Censo Escolar**: Estudantes de escolas públicas representam **70%** dos que trabalham enquanto estudam.

### Legislação Relevante

- **CLT, Art. 428**: Aprendizagem profissional limitada a 6 horas diárias.
- **Constituição Federal, Art. 7º, XXXIII**: Proibição de trabalho noturno, perigoso ou insalubre a menores de 18 anos.
- **Lei do Estágio (11.788/2008)**: Limite de 6 horas diárias para estudantes.
