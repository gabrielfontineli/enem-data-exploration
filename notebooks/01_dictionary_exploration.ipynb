# Data Dictionary Exploration - Enem 2023

This notebook explores the data dictionary provided with the Enem 2023 microdata to understand the variables and their schema.

```python
import pandas as pd
import os
import glob
```

## Load Dictionary File

We search for the dictionary file in the `data` directory.

```python
# Find the dictionary file using glob to handle potential special characters in path
base_path = '../data/interim/unzipped_2023/DICIONµRIO/'
# In case the path characters are regular in some environments or encoded differently, we try to match the pattern
search_pattern = os.path.join('../data/interim/unzipped_2023', '*', '*2023.xlsx')
files = glob.glob(search_pattern)

if not files:
    # Try a broader search if the specific path fails
    search_pattern = os.path.join('../data', '**', '*Microdados_Enem_2023.xlsx')
    files = glob.glob(search_pattern, recursive=True)

dict_path = files[0]
print(f"Found dictionary file: {dict_path}")
```

## Explore Sheets

```python
try:
    xl = pd.ExcelFile(dict_path)
    print(f"Sheet names: {xl.sheet_names}")
except ImportError as e:
    print("Error: openpyxl might be missing. Please install it using `pip install openpyxl`")
    raise e
```

## Content of Each Sheet

```python
for sheet in xl.sheet_names:
    print(f"\n--- Sheet: {sheet} ---")
    df = xl.parse(sheet)
    display(df.head())
    print(f"Shape: {df.shape}")
    print("Columns:", df.columns.tolist())
```
